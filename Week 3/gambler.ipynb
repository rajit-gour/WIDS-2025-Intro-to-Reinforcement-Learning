```python
states=[i for i in range(101)]
v=[0.0 for i in range(101)]
mdp={i:{} for i in states}
v[100]=1.0
ph=0.4
pt=0.6
theta=1e-9
```

```python
while True:
    delta=0
    vdash=v.copy()
    for s in range(1,100):
        vdash[s]=0.0
        for a in range(1,min(s,100-s)+1):
            vdash[s]=max(vdash[s],ph*v[s+a]+pt*v[s-a])
        delta=max(delta,abs(vdash[s]-v[s]))
    v=vdash
    if delta<theta:
        break
policy=[0]*101
for s in range(1,100):
    bestv=-1
    besta=0
    for a in range(1,min(s,100-s)+1):
        bla=ph*v[s+a]+pt*v[s-a]
        if bla>bestv:
            bestv=bla
            besta=a
    policy[s]=besta
```

```python
import matplotlib.pyplot as plt

plt.bar(range(1,100), policy[1:100])
plt.xlabel("Capital")
plt.ylabel("Stake")
plt.title("Optimal Policy for Gambler's Problem (p=0.4)")
plt.show()
```
